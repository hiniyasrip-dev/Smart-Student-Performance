# Smart ML System for Predictive Analysis of Student Performance

## 02 - Data Cleaning

This notebook focuses on cleaning and preparing the student performance dataset for exploratory data analysis, feature engineering, and machine learning.

The cleaning process includes:
- Checking and correcting data types
- Standardizing column names
- Checking missing and duplicate records
- Identifying invalid values
- Investigating derived variables
- Preventing data leakage
- Preparing the cleaned dataset for further analysis

```python
import pandas as pd
import numpy as np

from pathlib import Path
```

```python
file_path = Path("../data/raw/student_performance_analysis.xlsx")

df = pd.read_excel(
    file_path,
    sheet_name="Raw_Data"
)

df.head()
```

```python
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)

df.columns
```

```python
df = df.rename(columns={
    "perfomance_category": "performance_category",
    "attendence_category": "attendance_category"
})

df.columns
```

```python
df.dtypes
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
print("Negative study hours:")
print((df["weekly_self_study_hours"] < 0).sum())

print("\nInvalid attendance:")
print(((df["attendance_percentage"] < 0) | (df["attendance_percentage"] > 100)).sum())

print("\nInvalid participation:")
print(((df["class_participation"] < 0) | (df["class_participation"] > 100)).sum())

print("\nInvalid total scores:")
print(((df["total_score"] < 0) | (df["total_score"] > 100)).sum())
```

```python
pd.crosstab(
    df["grade"],
    pd.cut(
        df["total_score"],
        bins=[0, 50, 60, 70, 80, 90, 100],
        include_lowest=True
    )
)
```

```python
pd.crosstab(
    df["performance_category"],
    pd.cut(
        df["total_score"],
        bins=[0, 50, 60, 70, 80, 90, 100],
        include_lowest=True
    )
)
```

```python
df.groupby("study_hours_category")["weekly_self_study_hours"].agg(
    ["min", "max", "mean", "count"]
)
```

```python
df.groupby("attendance_category")["attendance_percentage"].agg(
    ["min", "max", "mean", "count"]
)
```

```python
df.groupby("participation_level")["class_participation"].agg(
    ["min", "max", "mean", "count"]
)
```

```python
model_df = df[
    [
        "weekly_self_study_hours",
        "attendance_percentage",
        "class_participation",
        "total_score"
    ]
].copy()

model_df.head()
```

```python
processed_path = Path("../data/processed")
processed_path.mkdir(parents=True, exist_ok=True)

model_df.to_csv(
    processed_path / "student_performance_model_data.csv",
    index=False
)

print("Modeling dataset saved successfully.")
```

## Data Cleaning Summary

The dataset was successfully cleaned and prepared for further analysis.

### Data Quality
- Total records: 5,010
- Missing values: None
- Original duplicate records: None
- Invalid numerical values: None

### Data Leakage Investigation
The following categorical variables were identified as derived variables:

- `grade` → derived from `total_score`
- `performance_category` → derived from `total_score`
- `study_hours_category` → derived from `weekly_self_study_hours`
- `attendance_category` → derived from `attendance_percentage`
- `participation_level` → derived from `class_participation`

These variables are retained in the original dataframe for descriptive analysis but are excluded from the initial machine learning feature set to prevent target leakage and redundant representation of numerical variables.

### Modeling Dataset

The initial regression model will use:

**Features**
- `weekly_self_study_hours`
- `attendance_percentage`
- `class_participation`

**Target**
- `total_score`

The modeling dataset has been saved as:

`data/processed/student_performance_model_data.csv`

This dataset will be used for exploratory data analysis, feature engineering, model training, and evaluation.
